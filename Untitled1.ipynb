```python
!pip install -q pypdf
from pypdf import PdfReader

reader = PdfReader("NIST.SP.800-53r5.pdf")

text = ""
for page in reader.pages:
    text += page.extract_text()
```

```python
def split_text(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap

    return chunks

chunks = split_text(text)
print("Total chunks:", len(chunks))
```

```python
from sentence_transformers import SentenceTransformer

embed_model = SentenceTransformer('all-MiniLM-L6-v2')

embeddings = embed_model.encode(chunks)
```

```python
!pip install -q faiss-cpu sentence-transformers
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

# Re-initialize the model and generate embeddings to ensure they are available
# (assuming 'chunks' is defined from earlier execution, as indicated by context)
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = embed_model.encode(chunks)

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings))
```

```python
if 'chunks' in locals() or 'chunks' in globals():
    print(f"'chunks' is defined. Type: {type(chunks)}, Length: {len(chunks)}")
else:
    print("'chunks' is not defined.")
```

```python
def retrieve(query, k=3):
    query_vector = embed_model.encode([query])

    distances, indices = index.search(query_vector, k)

    results = [chunks[i] for i in indices[0]]
    return results
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-base")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-base")
```

```python
query = "What is the purpose of NIST Special Publication 800-53?"
```

```python
relevant_chunks = retrieve(query)
```

```python
context = "\n".join(relevant_chunks)

prompt = f"Context: {context}\n\nQuestion: {query}\n\nAnswer:"

input_ids = tokenizer(prompt, return_tensors="pt").input_ids
outputs = model.generate(input_ids, max_length=512)
response = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(response)
```

```python
def rag(query):
    docs = retrieve(query)

    context = " ".join(docs)

    prompt = f"""
    Answer the question based only on the context below:

    Context:
    {context}

    Question:
    {query}
    """

    result = model.generate(tokenizer(prompt, return_tensors="pt").input_ids, max_length=512)
    answer = tokenizer.decode(result[0], skip_special_tokens=True)

    return answer, docs
```

```python
query = "Summarize the document"
answer = rag(query)

print("Answer:\n", answer)
```

```python
def agent(query):
    if "summarize" in query.lower():
        return rag(query)
    elif "explain" in query.lower():
        return rag(query)
    else:
        return rag(query)

print(agent("What is the main topic?"))
```

```python
chat_history = []

def agent_chat_memory():
    print("🤖 Smart Agent Ready! Type 'exit' to stop.\n")

    while True:
        query = input("You: ")

        if query.lower() == "exit":
            break

        chat_history.append(query)

        answer = rag(query)

        chat_history.append(answer)

        print("\nAgent:", answer)
        print("-" * 50)
```

```python
def agent_chat_ui():
    print("="*50)
    print("📄 Enterprise RAG Agent")
    print("="*50)

    while True:
        query = input("\n🧑 You: ")

        if query.lower() == "exit":
            print("\n🤖 Agent: Session ended.")
            break

        answer, references = rag(query)

        print("\n🤖 Agent:", answer)
        print("\n📚 References:")
        for i, ref in enumerate(references):
            print(f"  [{i+1}] {ref[:200]}...") # Print first 200 chars of each reference
```
